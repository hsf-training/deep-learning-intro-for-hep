## Automatic differentiation

PyTorch records operations on tensors that request gradients. Calling `backward` computes the derivative of the result with respect to those tensors.

```python
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
x.square().sum().backward()
print(x.grad)  # derivative of sum(x**2) is 2*x
```

## Fit a small model

This example trains a linear model to approximate `y = 2x + 1`. The data is deliberately small so it can run in a browser.

```python
torch.manual_seed(0)
features = torch.linspace(-1, 1, 40).reshape(-1, 1)
targets = 2 * features + 1
model = torch.nn.Linear(1, 1)
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)

for _ in range(120):
    optimizer.zero_grad()
    loss = torch.mean((model(features) - targets) ** 2)
    loss.backward()
    optimizer.step()

print('loss:', loss.item())
print('weight, bias:', model.weight.item(), model.bias.item())
```
